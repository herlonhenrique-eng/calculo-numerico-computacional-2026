```python
def f(x):
    return x**3 - x - 2


def bisseccao(f, a, b, n):
    for i in range(n):
        c = (a + b) / 2
        print("Iteração:", i, "C =", c)

        if f(a) * f(c) < 0:
            b = c
        else:
            a = c

    return (a + b) / 2
    # Chamando a função
raiz = bisseccao(f, 1, 2, 10)

print("A raiz aproximada é:", raiz)
```

O que aconteceu?

Ao colocar o primeiro código até a linha return (a + b) / 2, nada é exibido na tela. Isso ocorre porque a função foi somente definida, mas ainda não foi executada. Em outras palavras, o programa já tem as instruções, porém falta o comando que faz a função entrar em ação.

Quando utilizamos o print, a função é chamada, o cálculo é realizado e o resultado passa a aparecer na tela, mostrando uma aproximação da raiz da equação.

```python
def f(x):
    return x**3 - x - 2


def bisseccao(f, a, b, n):

    print("Iter |   a   |   b   |   c   |  f(c)")
    print("--------------------------------------")

    for i in range(n):

        c = (a + b) / 2

        print(i, "|", round(a,6), "|", round(b,6), "|", round(c,6), "|", round(f(c),6))

        if f(a) * f(c) < 0:
            b = c
        else:
            a = c

    return (a + b) / 2


raiz = bisseccao(f, 1, 2, 10)

print("\nRaiz aproximada:", raiz)
bisseccao(f,1,2,5)

bisseccao(f,1,2,10)

bisseccao(f,1,2,20)
```

O que acontece com o resultado?

Ao aumentar a quantidade de iterações no método da bisseção, o intervalo vai sendo refinado repetidamente, o que melhora a aproximação da raiz verdadeira. Assim, com 20 iterações, a resposta fica mais exata do que com 5 ou 10.
